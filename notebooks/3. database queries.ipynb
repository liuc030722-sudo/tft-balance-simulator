```python
import pandas as pd
from IPython.display import display

from tft_simulator.database.queries import (
    read_connection,
    list_versions,
    list_champions,
    get_champion,
    get_ability,
    get_counts,
)

with read_connection() as connection:
    versions = list_versions(connection)
versions_df = pd.DataFrame(versions)

display(
    versions_df[
        ["version_id", "set_number", "patch", "retrieved_at_utc"]
    ]
)

VERSION_ID = 1
```

```python
with read_connection() as connection:
    all_champions = list_champions(
        connection,
        version_id=VERSION_ID,
    )

    # cost=2表示只返回二费英雄。
    two_cost_champions = list_champions(
        connection,
        version_id=VERSION_ID,
        cost=2,
    )

print("当前版本英雄数量：", len(all_champions))

two_cost_df = pd.DataFrame(two_cost_champions)
display(
    two_cost_df[
        ["name", "cost", "hp", "attack_damage", "ability_name"]
    ]
)
```

```python
with read_connection() as connection:
    champion = get_champion(
        connection,
        version_id=VERSION_ID,
        champion_id="TFT10_Annie",
    )

    ability = get_ability(
        connection,
        version_id=VERSION_ID,
        champion_id="TFT10_Annie",
    )

# 找不到英雄时返回None
if champion is None:
    print("没有找到该英雄。")
else:
    print("英雄名称：", champion["name"])
    print("生命值：", champion["hp"])
    print("攻击力：", champion["attack_damage"])
    print("羁绊：", champion["traits"])
    
if ability is not None:
    print("技能名称：", ability["ability_name"])
    parameter_map = {
        parameter["name"]: parameter["value"]
        for parameter in ability["variables"]
    }
    print("主伤害原始参数：", parameter_map["MagicDamage"])
```
